```python
from qdrant_client import QdrantClient
from qdrant_client.models import VectorParams, Distance, PointStruct

client = QdrantClient(url="http://localhost:6333")
client.get_collections()
```

```python
client.create_collection(
    collection_name="toy_docs",
    vectors_config=VectorParams(size=4, distance=Distance.COSINE),
)
client.get_collections()
```

```python
client.delete_collection(collection_name="toy_docs")
```

```python
points = [
    PointStruct(id=1, vector=[0.9, 0.1, 0.0, 0.0], payload={"category": "fruit", "lang": "en"}),
    PointStruct(id=2, vector=[0.8, 0.2, 0.0, 0.0], payload={"category": "fruit", "lang": "id"}),
    PointStruct(id=3, vector=[0.0, 0.0, 0.9, 0.1], payload={"category": "vehicle", "lang": "en"}),
    PointStruct(id=4, vector=[0.4, 0.2, 0.8, 0.2], payload={"category": "vehicle", "lang": "id"}),
]

client.upsert(collection_name="toy_docs", points=points)
client.count(collection_name="toy_docs")
```

```python
query_vector=[0.85, 0.15, 0.0, 0.0]

results = client.query_points(
    collection_name="toy_docs",
    query=query_vector,
    limit=3,
    with_payload=True,
)
results.points
```

```python

from qdrant_client.models import Filter, FieldCondition, MatchValue, QueryResponse
```

```python

results: QueryResponse = client.query_points(
    collection_name="toy_docs",
    query=query_vector,
    query_filter=Filter(
        must=[FieldCondition(key="lang", match=MatchValue(value="id"))]
    ),
    limit=3,
    with_payload=True,
)

results.points
```

```python
vector_close_to_vehicle =[0.2, 0.3, 0.8, 0.3] 

results: QueryResponse = client.query_points(
  collection_name="toy_docs",
  query=vector_close_to_vehicle,
  query_filter=Filter(
    must_not=[FieldCondition(key="lang", match=MatchValue(value="en"))]
  ),
  limit=3,
  with_payload=True,
)

results.points
```

```python
from qdrant_client.models import PayloadSchemaType
```

```python
client.create_payload_index(
  collection_name="toy_docs",
  field_name="lang",
  field_schema=PayloadSchemaType.KEYWORD
)

client.get_collection("toy_docs")
```

```python
client.upsert(collection_name="toy_docs", points=[PointStruct(vector=[0.2, 0.5, 0.5, 0.4],id=5, payload={"category": "insect"})])
```

```python
client.facet(collection_name="toy_docs", key="lang")
```
